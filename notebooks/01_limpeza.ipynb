# Limpeza dos dados

Removi 3 colunas (EmployeeCount, Over18 e StandardHours) porque cada uma tem um único valor nas 1470 linhas e não ajuda a explicar a saída. Criei 2 colunas novas: AttritionFlag (1 para quem saiu, 0 para quem ficou) e FaixaEtaria. Resultado: 1470 linhas e 34 colunas, sem valores nulos.

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.nunique()
```

```python
df["StandardHours"].nunique()
```

```python
df["StandardHours"].value_counts()
```

```python
df = df.drop(columns=["EmployeeCount","Over18","StandardHours"])
```

```python
df.shape
```

```python
df["AttritionFlag"] = (df["Attrition"] == 'Yes').astype(int)
```

```python
df["AttritionFlag"].sum()
```

```python
df["AttritionFlag"].mean()
```

```python
df["FaixaEtaria"] = pd.cut(df["Age"], bins=[17, 25, 35, 45, 60], labels=["18-25", "26-35", "36-45", "46-60"])
```

```python
df["FaixaEtaria"].value_counts()
```

```python
df.to_csv("../data/processed/employees_clean.csv", index=False)
```
